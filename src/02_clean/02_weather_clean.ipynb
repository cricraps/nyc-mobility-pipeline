```python
%sql
CREATE OR REPLACE TABLE nyc_mobility.clean.weather_silver AS
WITH cleaned_weather AS (
    SELECT
    -- correct data types based on data profiling
        elevation,
        latitude,
        longitude,
        observation_time,
        precipitation,
        rain,
        snowfall,
        temperature_2m,
        TRIM(timezone) AS timezone,
        CAST(weather_code AS INT) AS weather_code,
        wind_speed_10m

    FROM nyc_mobility.raw.weather
)

SELECT DISTINCT *
FROM cleaned_weather
WHERE observation_time IS NOT NULL;
```
