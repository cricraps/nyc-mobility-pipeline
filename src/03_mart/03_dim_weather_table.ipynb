```python
%sql

-- WEATHER DIMENSION TABLE --
-- Grain: 1 row per date and hour (city-wide NYC)
-- Note: precipitation is not always >= rain + snowfall in source data;
--   downstream weather-condition logic should check snowfall before precipitation

CREATE OR REPLACE TABLE nyc_mobility.mart.dim_weather AS
SELECT 
    CAST(date_format(observation_time, 'yyyyMMddHH') AS INT) AS date_hour_key,
    CAST(observation_time AS DATE) AS observation_date,
    HOUR(observation_time) AS observation_hour,
    temperature_2m,
    precipitation,
    rain,
    snowfall,
    weather_code,
    wind_speed_10m
FROM nyc_mobility.clean.weather_silver
WHERE observation_time IS NOT NULL;
```
