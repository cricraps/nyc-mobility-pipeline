```python
%sql

-- ADVISORY DIMENSION TABLE --
-- Grain: 1 row per traffic advisory, scoped by borough 

CREATE OR REPLACE TABLE nyc_mobility.mart.dim_advisory AS
SELECT 
    closure_bk AS advisory_id,
    section_id,
    entry_type AS advisory_type,
    advisory_text AS description,
    borough,
    location_name,
    street_name,
    from_street,
    to_street,
    COALESCE(effective_from, advisory_week_start) AS effective_from,
    COALESCE(effective_to, advisory_week_end) AS effective_to
FROM nyc_mobility.clean.traffic_advisory
WHERE closure_bk IS NOT NULL;
```
