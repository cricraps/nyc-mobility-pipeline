```python
%sql

-- ADVISORY DIMENSION TABLE --

CREATE OR REPLACE TABLE nyc_mobility.mart.dim_advisory AS
SELECT 
    section_id AS advisory_id,
    section_name,
    borough,
    location_name,
    street_name,
    from_street,
    to_street,
    entry_type,
    effective_from,
    effective_to
FROM nyc_mobility.clean.traffic_advisory
WHERE section_id IS NOT NULL;
```

```python
%sql

-- DATE DIMENSION TABLE --

CREATE OR REPLACE TABLE nyc_mobility.mart.dim_date AS
SELECT DISTINCT
    CAST(date_format(lpep_pickup_datetime, 'yyyyMMdd') AS INT) AS date_key,
    CAST(lpep_pickup_datetime AS DATE) AS full_date,
    YEAR(lpep_pickup_datetime) AS year,
    MONTH(lpep_pickup_datetime) AS month,
    DAY(lpep_pickup_datetime) AS day,
    DATE_FORMAT(lpep_pickup_datetime, 'EEEE') AS day_of_week,
    CASE 
        WHEN dayofweek(lpep_pickup_datetime) IN (1, 7) THEN TRUE 
        ELSE FALSE 
    END AS is_weekend
FROM nyc_mobility.clean.green_taxi
WHERE lpep_pickup_datetime IS NOT NULL;
```

```python
%sql

-- WEATHER DIMENSION TABLE --

CREATE OR REPLACE TABLE nyc_mobility.mart.dim_weather AS
SELECT 
    CAST(date_format(observation_time, 'yyyyMMdd') AS INT) AS date_key,
    HOUR(observation_time) AS hour_of_day,
    temperature_2m,
    precipitation,
    rain,
    snowfall,
    weather_code,
    wind_speed_10m
FROM nyc_mobility.clean.weather_silver
WHERE observation_time IS NOT NULL;
```

```python
%sql

-- ZONE DIMENSION TABLE --

CREATE OR REPLACE TABLE nyc_mobility.mart.dim_zone AS
SELECT 
    location_id,
    borough,
    zone,
    service_zone
FROM nyc_mobility.clean.taxi_zones
WHERE location_id IS NOT NULL;
```
