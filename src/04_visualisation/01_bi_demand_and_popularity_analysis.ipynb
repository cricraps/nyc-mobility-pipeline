```python
%sql

-- Question 1a: Which borough-to-borough flows carry the highest trip volume, distance, and revenue?
-- Suggested chart: Heatmap | Rows: pickup_borough | Columns: dropoff_borough | Color intensity: total_trips

SELECT 
    pz.borough AS pickup_borough,
    dz.borough AS dropoff_borough,
    COUNT(t.trip_id) AS total_trips,
    ROUND(SUM(t.trip_distance), 2) AS total_distance_miles,
    ROUND(SUM(t.total_amount), 2) AS total_revenue
FROM nyc_mobility.mart.fact_trip AS t
JOIN nyc_mobility.mart.dim_zone AS pz 
    ON t.pu_location_id = pz.location_id
JOIN nyc_mobility.mart.dim_zone AS dz 
    ON t.do_location_id = dz.location_id
GROUP BY pickup_borough, dropoff_borough
ORDER BY total_trips DESC;
```

```python
%sql

-- Question 1b: Which individual pickup zones drive the most trip volume and revenue?
-- Suggested chart: Horizontal bar | X: total_trips | Y: pickup_zone (top 10) | Color: pickup_borough

SELECT 
    pz.borough AS pickup_borough,
    pz.zone AS pickup_zone,
    COUNT(t.trip_id) AS total_trips,
    ROUND(AVG(t.trip_distance), 2) AS avg_trip_distance_miles,
    ROUND(SUM(t.total_amount), 2) AS total_revenue
FROM nyc_mobility.mart.fact_trip AS t
JOIN nyc_mobility.mart.dim_zone AS pz 
    ON t.pu_location_id = pz.location_id
GROUP BY pickup_borough, pickup_zone
ORDER BY total_trips DESC
LIMIT 10;
```
