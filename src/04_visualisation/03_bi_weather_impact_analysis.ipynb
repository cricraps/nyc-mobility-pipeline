```python
%sql

-- Question 3: How do weather conditions (rain, snow, clear) impact trip demand, distance, and duration?
-- Suggested chart: Combo (bar + line) | X: weather_condition | Bar (left Y): total_trips | Line (right Y): avg_duration_minutes

SELECT 
    CASE 
        WHEN w.precipitation > 0 THEN 'Rain/Precipitation'
        WHEN w.snowfall > 0 THEN 'Snow'
        ELSE 'Clear/Dry'
    END AS weather_condition,
    ROUND(AVG(w.temperature_2m), 1) AS avg_temperature_c,
    COUNT(t.trip_id) AS total_trips,
    ROUND(AVG(t.trip_distance), 2) AS avg_distance_miles,
    ROUND(AVG(t.trip_duration_min), 2) AS avg_duration_minutes
FROM nyc_mobility.mart.fact_trip AS t
JOIN nyc_mobility.mart.dim_weather AS w 
    ON t.pickup_date_hour_key = w.date_hour_key
GROUP BY weather_condition
ORDER BY total_trips DESC;
```
