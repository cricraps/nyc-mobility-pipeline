```python
%sql

-- Question 2: How do passenger volume, fare amounts, and trip durations vary across hours of the day, days of the week, and weekends vs. weekdays?

SELECT 
    d.day_of_week,
    d.is_weekend,
    t.pickup_hour,
    COUNT(t.trip_id) AS total_trips,
    SUM(t.passenger_count) AS total_passengers,
    ROUND(AVG(t.trip_duration_min), 2) AS avg_duration_minutes,
    ROUND(AVG(t.fare_amount), 2) AS avg_fare_amount,
    ROUND(AVG(t.total_amount), 2) AS avg_total_amount
FROM nyc_mobility.mart.fact_trip AS t
JOIN nyc_mobility.mart.dim_date AS d 
    ON t.pickup_date_key = d.date_key
GROUP BY 
    d.day_of_week, 
    d.is_weekend, 
    t.pickup_hour
ORDER BY 
    d.is_weekend,
    t.pickup_hour;
```
