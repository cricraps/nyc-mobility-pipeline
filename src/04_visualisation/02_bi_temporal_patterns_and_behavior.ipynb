```python
%sql

-- Question 2: How do passenger volume, fare amounts, and trip durations vary across hours of the day and weekends vs. weekdays?
-- Suggested chart: Line chart | X: pickup_hour | Y: total_trips | Series/Color: is_weekend

CREATE OR REPLACE TABLE nyc_mobility.bi_visualization.temporal_patterns_and_behaviors AS
SELECT 
    t.pickup_hour,
    d.is_weekend,
    COUNT(*) AS total_trips,
    SUM(t.passenger_count) AS total_passengers,
    ROUND(AVG(t.trip_duration_min), 2) AS avg_duration_minutes,
    ROUND(AVG(t.fare_amount), 2) AS avg_fare_amount
FROM nyc_mobility.mart.fact_trip AS t
JOIN nyc_mobility.mart.dim_date AS d 
    ON t.pickup_date_hour_key = d.date_hour_key
GROUP BY t.pickup_hour, d.is_weekend
ORDER BY d.is_weekend, t.pickup_hour;
```
