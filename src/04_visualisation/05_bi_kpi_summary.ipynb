```python
%sql

-- Consolidated KPI Summary Banner
CREATE OR REPLACE TABLE nyc_mobility.bi_visualization.kpi_summary AS
SELECT 
    COUNT(*) AS total_trips,
    SUM(passenger_count) AS total_passengers,
    ROUND(SUM(total_amount), 2) AS total_revenue_usd,
    ROUND(AVG(fare_amount), 2) AS avg_fare_usd,
    ROUND(AVG(trip_distance), 2) AS avg_distance_miles,
    ROUND(AVG(trip_duration_min), 2) AS avg_duration_minutes,
    ROUND(SUM(total_amount) / NULLIF(SUM(trip_distance), 0), 2) AS revenue_per_mile
FROM nyc_mobility.mart.fact_trip;
```
