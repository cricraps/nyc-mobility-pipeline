```python
%sql
CREATE OR REPLACE VIEW nyc_mobility.validation.clean_layer_quality AS

SELECT
    'Green Taxi' AS dataset,
    `column`,
    data_quality_check,
    failed_rows,
    total_rows,
    percentage,
    status
FROM nyc_mobility.validation.green_taxi_validation

UNION ALL

SELECT
    'Weather' AS dataset,
    `Column` AS `column`,
    data_quality_check,
    failed_rows,
    total_rows,
    `Percentage` AS percentage,
    status
FROM nyc_mobility.validation.weather_validation

UNION ALL

SELECT
    'Taxi Zones' AS dataset,
    `column`,
    data_quality_check,
    failed_rows,
    total_rows,
    percentage,
    status
FROM nyc_mobility.validation.taxi_zones_validation

UNION ALL

SELECT
    'Traffic Advisory' AS dataset,
    `column`,
    data_quality_check,
    failed_rows,
    total_rows,
    percentage,
    status
FROM nyc_mobility.validation.traffic_advisory_validation;
```

```python
%sql
SELECT *

FROM nyc_mobility.validation.clean_layer_quality

ORDER BY dataset, data_quality_check;
```
